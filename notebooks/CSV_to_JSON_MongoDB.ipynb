# **Xử lý dữ liệu Orders**

```python
# Tạo bản copy để xử lý
orders_json = orders.copy()
items_json = items.copy()
payments_json = payments.copy()
reviews_json = reviews.copy()
```

```python
# Chuẩn hóa các cột ngày giờ
# Orders
order_date_cols = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in order_date_cols:
    orders_json[col] = pd.to_datetime(
        orders_json[col],
        errors="coerce"
    ).dt.strftime("%Y-%m-%dT%H:%M:%S")


# Order Items
items_json["shipping_limit_date"] = pd.to_datetime(
    items_json["shipping_limit_date"],
    errors="coerce"
).dt.strftime("%Y-%m-%dT%H:%M:%S")


# Reviews
review_date_cols = [
    "review_creation_date",
    "review_answer_timestamp"
]

for col in review_date_cols:
    reviews_json[col] = pd.to_datetime(
        reviews_json[col],
        errors="coerce"
    ).dt.strftime("%Y-%m-%dT%H:%M:%S")
```

```python
# Chuyển NaN thành None
orders_json = orders_json.astype(object).where(
    pd.notnull(orders_json), None
)

items_json = items_json.astype(object).where(
    pd.notnull(items_json), None
)

payments_json = payments_json.astype(object).where(
    pd.notnull(payments_json), None
)

reviews_json = reviews_json.astype(object).where(
    pd.notnull(reviews_json), None
)
```

```python
# Gom items theo từng order_id
items_grouped = {
    order_id: group.drop(columns="order_id").to_dict("records")
    for order_id, group in items_json.groupby("order_id")
}
```

```python
len(items_grouped)
```

```python
# Gom payments theo order
payments_grouped = {
    order_id: group.drop(columns="order_id").to_dict("records")
    for order_id, group in payments_json.groupby("order_id")
}
```

```python
len(payments_grouped)
```

```python
# Gom reviews
reviews_grouped = {
    order_id: group.drop(columns="order_id").to_dict("records")
    for order_id, group in reviews_json.groupby("order_id")
}
```

```python
len(reviews_grouped)
```

```python
# Tạo nested document cho từng order
order_documents = []

for order in orders_json.to_dict("records"):

    order_id = order["order_id"]

    order["items"] = items_grouped.get(order_id, [])
    order["payments"] = payments_grouped.get(order_id, [])
    order["reviews"] = reviews_grouped.get(order_id, [])

    order_documents.append(order)
```

```python
len(order_documents)
```

```python
# Xem document thật trước khi xuất
import json

print(
    json.dumps(
        order_documents[0],
        indent=2,
        ensure_ascii=False
    )
)
```

# **Xử lý và xuất Geolocation sang JSON**

```python
# Khảo sát Geolocation trước
print("Rows:", len(geo))

print(
    "Unique ZIP prefixes:",
    geo["geolocation_zip_code_prefix"].nunique()
)

print(
    "Duplicate rows:",
    geo.duplicated().sum()
)

print("\nMissing values:")
print(geo.isnull().sum())
```

```python
geo.groupby(
    "geolocation_zip_code_prefix"
).size().sort_values(ascending=False).head(10)
```

```python
geo.head()
```

```python
# Xóa duplicate hoàn toàn
geo_clean = geo.drop_duplicates().copy()

print("Before:", len(geo))
print("After removing duplicates:", len(geo_clean))
```

```python
# Aggregate theo ZIP prefix
def get_mode(series):
    mode = series.mode()
    return mode.iloc[0] if not mode.empty else None


geo_agg = (
    geo_clean
    .groupby("geolocation_zip_code_prefix")
    .agg(
        geolocation_lat=("geolocation_lat", "median"),
        geolocation_lng=("geolocation_lng", "median"),
        geolocation_city=("geolocation_city", get_mode),
        geolocation_state=("geolocation_state", get_mode)
    )
    .reset_index()
)
```

```python
# Kiểm tra kết quả
print("Rows:", len(geo_agg))
print(
    "Unique ZIP:",
    geo_agg["geolocation_zip_code_prefix"].nunique()
)

print("\nMissing:")
print(geo_agg.isnull().sum())

geo_agg.head()
```

```python
# Kiểm tra khả năng JOIN với customers và sellers
customer_zip_coverage = customers[
    "customer_zip_code_prefix"
].isin(
    geo_agg["geolocation_zip_code_prefix"]
).mean()

seller_zip_coverage = sellers[
    "seller_zip_code_prefix"
].isin(
    geo_agg["geolocation_zip_code_prefix"]
).mean()

print(f"Customer ZIP coverage: {customer_zip_coverage:.2%}")
print(f"Seller ZIP coverage: {seller_zip_coverage:.2%}")
```

```python
# Xuất geolocation.json
geo_json = geo_agg.astype(object).where(
    pd.notnull(geo_agg),
    None
)

geo_documents = geo_json.to_dict("records")

print("Number of geolocation documents:", len(geo_documents))
```

# **Kết nối MongoDB Atlas**

```python
# Kết nối MongoDB Atlas

import os
from pathlib import Path
from dotenv import load_dotenv
from pymongo import MongoClient

# Xác định thư mục gốc project
PROJECT_ROOT = Path(r"D:\olist-ecommerce-end-to-end-analytics")

# Đọc file .env
env_path = PROJECT_ROOT / ".env"
load_dotenv(env_path, override=True)

# Kiểm tra cấu hình
print("File .env tồn tại:", env_path.exists())
print("MongoDB URI đã cấu hình:", bool(os.getenv("MONGODB_URI")))

# Kết nối MongoDB
uri = os.getenv("MONGODB_URI")

if not uri:
    raise ValueError("Không tìm thấy MONGODB_URI trong file .env")

client = MongoClient(uri, serverSelectionTimeoutMS=10000)

try:
    client.admin.command("ping")
    print("Kết nối MongoDB Atlas thành công!")
except Exception as e:
    print("Kết nối thất bại:", type(e).__name__)
    print(str(e))
```

## **Import dữ liệu vào MongoDB**

```python
# Import JSON vào MongoDB Atlas

import json

db = client["olist_db"]

collection_names = [
    "customers",
    "geolocation",
    "orders",
    "products",
    "sellers"
]

batch_size = 1000

for name in collection_names:
    collection = db[name]
    current_count = collection.count_documents({})

    if current_count > 0:
        print(f"{name}: Already contains {current_count:,} documents. Skipped.")
        continue

    file_path = PROJECT_ROOT / "data" / "json" / f"{name}.json"

    with open(file_path, "r", encoding="utf-8") as f:
        documents = json.load(f)

    if not isinstance(documents, list):
        raise ValueError(f"{name}.json must contain a JSON array")

    if not documents:
        print(f"{name}: Empty JSON file. Skipped.")
        continue

    for i in range(0, len(documents), batch_size):
        batch = documents[i:i + batch_size]
        collection.insert_many(batch)

    print(f"{name}: Imported {collection.count_documents({}):,} documents")
```

```python
# Create MongoDB indexes

indexes = {
    "orders": "order_id",
    "customers": "customer_id",
    "products": "product_id",
    "sellers": "seller_id"
}

for collection_name, field in indexes.items():
    index_name = db[collection_name].create_index(
        field,
        unique=True
    )
    print(f"{collection_name}: {index_name}")

print("\nMongoDB indexing completed!")
```
